```python
import pandas as pd
```

```python
# Read the TSV file properly by handling the wrapping double quotes.
# We can use pd.read_csv with custom configurations or manually strip quotes.
with open("marketing_campaign.csv", "r") as f:
    lines = [line.strip().strip('"') for line in f.readlines()]

# Create DataFrame from parsed rows
data = [line.split("\t") for line in lines]
df = pd.DataFrame(data[1:], columns=data[0])

# Convert numerical columns to correct types
for col in df.columns:
    df[col] = pd.to_numeric(df[col], errors='ignore')

# Let's inspect columns to verify they loaded properly
print("Columns:", list(df.columns))
print("Shape:", df.shape)
```

```python
df = df.drop_duplicates()
```

```python
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)
```

```python
# Standardize text values
df["education"] = df["education"].str.strip().str.title()
df["marital_status"] = df["marital_status"].str.strip().str.title()
```

```python
# Convert date column
df["dt_customer"] = pd.to_datetime(
    df["dt_customer"],
    format="%d-%m-%Y",
    errors="coerce"
)
```

```python
 #Check data types
print(df.dtypes)

# Export cleaned dataset
df.to_csv("cleaned_customer_personality.csv", index=False)

print("Data cleaning completed successfully!")
```
